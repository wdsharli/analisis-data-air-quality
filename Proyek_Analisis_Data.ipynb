## Menentukan Pertanyaan Bisnis

**SMART Question** adalah sebuah framework untuk merumuskan pertanyaan secara terstruktur agar memeperoleh informasi yang mendalam.

* **Spesific (Spesifik)**
  * Pertanyaan harus jelas, fokus pada sebuah topik tertentu, dan tidak bermakna ganda. Hindari pertanyaan yang terlalu luas.
    * Salah: Bagaimana cara meningkatkan penjualan?
    * Benar: Faktor apa saja yang memengaruhi penurunan penjualan produk kategori elektronik di wilayah Jakarta selama kuartal terakhir?
* **Measurable (Terukur)**
  * Pertanyaan harus bisa dijawab dengan angka atau matrix yang konkret, harus tahu apa yang akan dihitung.
    * Salah: Apakah pelanggan senang dengan layanan kita?
    * Benar: Berapa skor rata-rata Customer Satisfaction untuk layanan purna jual bulan ini dibandingkan bulan lalu?
* **Action-Oriented (Berorientasi Aksi)**
  * Hasil dari pertanyaan harus bisa memberikan arahan untuk melakukan tindakan nyata. Jika pertanyaan terjawab, stakeholder harus tahu apa langkah selanjutnya.
    * Salah: Mengapa orang suka berbelanja?
    * Benar: Fitur apa pada aplikasi yang paling sering digunakan sebelum pengguna memutuskan untuk melakukan checkout?
* **Relevant (Relevan)**
  * Hasil dari pertanyaan harus sejalan dengan tujuan utama bisnis atau masalah yang sedang dihadapi.
    * Salah: Menanyakan tentang stok gudang saat masalah utamanya adalah efektivitas kampanye media sosial.
    * Benar: Apakah kampanye iklan di Instagram memberikan Return on Ad Spend (ROAS) yang lebih tinggi dibandingkan iklan di TikTok?
* **Time-bound (Terikat Waktu)**
  * Pertanyaan harus ada batasan waktu yang jelas agar analisis memiliki konteks yang tepat.
    * Salah: Berapa banyak pengguna baru kita?
    * Benar: Berapa tingkat pertumbuhan pengguna baru secara bulanan (Month-over-Month) sepanjang tahun 2025?

**Contoh pertanyaan bisnis yang memenuhi seluruh elemen SMART**

***"Faktor apa saja yang memengaruhi penurunan conversion rate pada pengguna aplikasi Android di wilayah Jabodetabek sebesar 5% selama periode Flash Sale Maret 2026?"***

Keterangan:

- **Specific**: Fokus pada "penurunan conversion rate" untuk "aplikasi Android" di "Jabodetabek". Bukan sekadar penjualan turun.
- **Measurable**: Ada angka konkret yang ingin dianalisis, yaitu penurunan sebesar "5%".
- **Action-Oriented**: Dengan mengetahui faktor penyebabnya misalnya bug pada versi Android tertentu atau kendala logistik di Jabodetabek, tim bisa langsung melakukan perbaikan teknis atau operasional.
- **Relevant**: Penurunan konversi saat Flash Sale adalah masalah kritis bagi bisnis retail/e-commerce.
- **Time-bound**: Dibatasi pada periode spesifik "Flash Sale Maret 2026".

## Import Semua Packages/Library yang Digunakan

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## Data Wrangling

### Gathering Data

#### Load df ...

```python
#Memuat dataset ke dalam DataFrame
df = pd.read_csv("PRSA_Data_Aotizhongxin_20130301-20170228.csv")

#Menampilkan 5 baris pertama data
df.head()
```

**Insight:** (Opsional)
- Berhasil memuat dataset stasiun Aotizhongxin
- Kolom data mencakup waktu (year, month, day, hour) dan berbagai parameter polutan (PM2.5, PM10, SO2, CO, O3), kondisi meteorologi (suhu, tekanan, titik embun, curah hujan, arah dan kecepatan angin), serta nama stasiun.

### Assessing Data

#### Identifying ... problem

```python
#Melihat informasi umum tentang data
print("INFO DATA")
df.info()

#Mengecek jumlah Missing Value
print("Missing Value")
print(df.isna().sum())

#Mengecek Jumlah data yang duplikat
print("JUMLAH DATA DUPLIKAT")
print(df.duplicated().sum())
```

**Steps to Take:**
- Menggunakan metode info() untuk melihat jumlah baris, kolom, dan tipe data
- Menggunakan fungsi isna().sum() untuk mendeteksi jumlah missing value pada setiap kolom.
- Mengecek duplikasi data menggunakan duplicated().sum().

**Insight:** (Opsional)
- Tidak terdapat duplikat data
- Ditemukan adanya missing value pada kolom polutan dan meteorologi.

### Cleaning Data

#### Fixing ... problem

```python
#Mengisi missing vakue menggunakan metode interpolasi
df_cleaned = df.interpolate(method='linear', axis=0)

#Untuk kolom kategori teks seperti arah angin (wd), isi dengan nilai yang sering muncul(modus)
if 'wd' in df_cleaned.columns:
  df_cleaned['wd'].fillna(df_cleaned['wd'].mode()[0], inplace=True)

#Mengecek sisa missing value
print(df_cleaned.isna().sum())
```

**Insight:** (Opsional)
- Metode Interpolasi sangat bagus untuk data time-series seperti pengamatan polusi udara ini karena memperkirakan nilai berdasarkan tren data sebelum dan sesudahnya
- Metode interpolasi dan modus berhasil mengatasi seluruh missing value pada dataset

## Exploratory Data Analysis (EDA)

### Explore ...

```python
#Mengelompokan data berdasarkan tahun dan menghitung rata-rata PM2.5
pm25_yearly = df_cleaned.groupby('year')['PM2.5'].mean().reset_index()

#Menampilkan hasil pengelompokan
pm25_yearly

```

**Insight:** (Opsional)


## Visualization & Explanatory Analysis

### Pertanyaan 1:

```python
#Membuat visualisasi Line Chart
plt.figure(figsize=(10, 5))
plt.plot(pm25_yearly['year'], pm25_yearly['PM2.5'], marker='o', color='b', linewidth=2)
plt.title('Tren rata-rata konsentrasi PM2.5 per Tahun (Stasiun Aotizhongxin)', fontsize=14)
plt.xlabel('Tahun', fontsize=12)
plt.ylabel(r'Rata-rata PM2.5 ($\mu g/m^3$)', fontsize=12)
plt.grid(True)
plt.show()
```

### Pertanyaan 2:

```python
#Mengelompokan data kecepatan angin (WSPM) dan melihat rata-rata PM10
wspm_pm10 = df_cleaned.groupby('WSPM')['PM10'].mean().reset_index()

#Membuat Visualisasi Scatter Plot dengan garis tren
plt.figure(figsize=(10,5))
sns.scatterplot(data=wspm_pm10, x='WSPM', y='PM10', color='r', alpha=0.7)
plt.title('Hubungan Kecepatan Angin (WSPM) terhadap Konsentrasi Rata-rata PM10', fontsize=14)
plt.xlabel('Kecepatan Angin (WSPM - m/s)', fontsize=12)
plt.ylabel(r'Rata-rata PM10($\mu g/m63$)', fontsize=12)
plt.grid(True)
plt.show
```

**Insight:** (Opsional)

**Pertanyaan 1**
- Rata-rata konsentrasi polutan PM2.5 naik pada tahun 2014, kemudian turun sampai tahun 2016, dan naik secara signifikan pada tahun 2017
Tahun 2017 menunjukan tingkat polusi tertinggi.
- Konsentrasi rata-rata polutan PM10 cenderung tinggi saat kecepatan angin rendah.

**Pertanyaan 2**
- Seiring meningkatnya kecepatan angin diatas 3m/s, konsentrasi rata-rata PM10 secara umum menurun dan lebih terdispersi. Hal ini menunjukan bahwa kecepatan angin membantu menyebarkan atau mengurangi penumpukan partikel polutan PM10 di udara sekitar stasiun Aotizhongxin.
